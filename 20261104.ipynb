```python
a = int(input('2 이상의 정수를 입력하세요'))
for i in range(2,a):
  if a%i==0:
    break
if a == i + 1:
  print(f'{a}는 소수 입니다.')
elif a == 2:
  print(f'{a}는 소수 입니다.')
else:
  print(f'{a}는 소수가 아닙니다.')
```

```python
num1 = int(input('정수를 입력하세요.'))
num2 = int(input('정수를 입력하세요.'))
b=min(num1,num2)
a=max(num1,num2)
for i in range(b,a+1):
  if i%3==1 or i%3==2:
    continue
  elif i%3==0:
    print(i)
```
