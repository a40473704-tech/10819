```python
def S1(num):
  if num % 2 == 0:
    return '짝수'
  else:
    return '홀수'

int_num = int(input('정수를 입력하세요'))
print(S1(int_num))
```

```python
def s1(base, height):
    area = (base* height) / 2
    return area
base = float(input("삼각형의 밑변을 입력하세요: "))
height = float(input("삼각형의 높이를 입력하세요: "))
result = s1(base, height)
print(f"밑변이 {base}, 높이가 {height}인 삼각형의 넓이는 {result}입니다.")
```
