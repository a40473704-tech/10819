```python
sentence1 = input('첫 번째 영어 문장을 입력하세요.')
sentence2 = input('두 번째 영어 문장을 입력하세요.')
sentence3 = input('세 번째 영어 문장을 입력하세요.')
text= sentence1+''+sentence2+''+sentence3
text=text.replase(',','').replase(',','')
text=text.lower()
words=text.split()
unicode_literals=set(words)
sorted_unicode_words=sorted(unicode_literals)
print('\n단어목록')
print(sorted_unicode_words)
```
